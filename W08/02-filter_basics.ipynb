This exercise is about understanding the fundamentals of linear filters to shift pixels in a certain direction. The actual filter may have little practical relevance.

 that returns a filter that shifts the pixels of an image by `n`

3. Create the functions `shift_to_left`
, `shift_to_up`
, and `shift_to_down`
 to shift the input images $n$ pixels to the left, up and down. Use one of the example images to test these functions and evaluate whether the images are correctly shifted. 
4. Modify the filter to not only shift the pixel to the right (or the other directions) but also decreases its brightness.


---

```python
def shift_to_right(n):
    """
    Param n: number of pixels to shift to the right.

    Returns: filter for shifting.
    """
    shift_filter = np.zeros((1,2*n+1)) # 2*n due to the filter choosing the middle value for the pixel
    shift_filter[0,2*n] = 1
    return shift_filter

#plt.imshow(data.coins(), cmap='gray')
#plt.imshow(convolve(data.coins(), shift_to_right(100)), cmap='gray')

def shift_to_left(n):
    """
    Param n: number of pixels to shift to the left.

    Returns: filter for shifting.
    """
    shift_filter = np.zeros((1,2*n+1)) # 2*n due to the filter choosing the middle value for the pixel
    shift_filter[0,0] = 1
    return shift_filter

def shift_to_up(n):
    """
    Param n: number of pixels to shift upwards.

    Returns: filter for shifting.
    """
    shift_filter = np.zeros((2*n+1,1)) # 2*n due to the filter choosing the middle value for the pixel
    shift_filter[0,0] = 1
    return shift_filter

def shift_to_down(n):
    """
    Param n: number of pixels to shift downwards.

    Returns: filter for shifting.
    """
    shift_filter = np.zeros((2*n+1,1)) # 2*n due to the filter choosing the middle value for the pixel
    shift_filter[2*n,0] = 1
    return shift_filter

def shift_to_down_and_lower_brightness(n):
    """
    Param n: number of pixels to shift downwards.

    Returns: filter for shifting.
    """
    shift_filter = np.zeros((2*n+1,1)) # 2*n due to the filter choosing the middle value for the pixel
    shift_filter[2*n,0] = 0.03 # decreased brightness
    return shift_filter


plt.title("shifted down 100 pixels")
plt.imshow(convolve(data.coins(), shift_to_down(100)), cmap='gray')
plt.show()

plt.title("shifted right 100 pixels")
plt.imshow(convolve(data.coins(), shift_to_right(100)), cmap='gray')
plt.show()

plt.figure()
plt.title("shifted down 100 pixels, and lowered brightness")
plt.imshow(convolve(data.coins(), shift_to_down_and_lower_brightness(100)), cmap='gray')
plt.show()
```
