2. Each of the exercises above are designed to have a single exact solution. How does assuming a single exact solution relate to real world problems? Consider inputs and labels as part of your argument. 

3. In [Task 3](#weight2), what would happen, if more than three training points were available?

4. What would happen to the solutions if the training points have small errors (noise)?

```python
# -------------- 1 --------------
# These exercises show how fitting a model to data means choosing weights 
# so the model gives the correct output for the given input points.

# -------------- 2 --------------
# Assuming that we have a single exact solution means inputs and labels are treated
# as "noise-free" and perfectly consistent. In real-world problems, inputs 
# are noisy and labels are imperfect, so exact solutions rarely exist and we
# usually instead seek to approximate solutions.

# -------------- 3 ---------------
# With more than three training points in Task 3, the system would generally 
# become 'overdetermined', meaning we have more points then unknown parameters
# and therefore no exact solution exists and the weights would need to be 
# estimated using a criterion such as least squares.


# ------------- 4 ---------------
# Small errors (noise) in the training points would change the exact solution
# or even remove it entirely, making the estimated weights sensitive to noise.

# ------------- 5 ---------------
# In Task 4, models D and E are linear in the weights despite being nonlinear 
# in the inputs, while model F shows that linearity refers to how weights enter
# the model, not whether each weight appears only once or with simple input terms.
```
